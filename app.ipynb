```python
import json
from ultralytics import  YOLO
model = YOLO("yolo11n.pt")
```

```python
predictions = model.predict(source= "test_images/test_img01.png",
                            conf= 0.25,
                            save= True)
data = {}
for predicted in predictions:
    for predicted in predicted.boxes:
        conf = float(predicted.conf[0])
        class_index = int(predicted.cls[0])
        cordinates = predicted.xyxy[0].tolist()
        class_name = model.names[class_index]
        print(f"Object Name: {class_name}\nConfidence Score: {conf * 100:.2f}\n")
        data[class_name] = {"class index": class_index, "Confidence Score": round(conf, 2) * 100, "Corordinates": cordinates}
data_json = json.dumps(data)
print(data_json)
```
